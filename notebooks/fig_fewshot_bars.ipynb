# Few-shot support-scaling bar chart (replaces tab:fewshot)

PrecogRM (per-benchmark specialist, 3-seed mean) vs the strongest per-task
baseline at each support size $k$, per benchmark. Vertical 3-panel stack for
single-column placement; dashed line = baseline at $k{=}8$.
Numbers from `support_sweep.json` aggregation (same as the old table).
Writes `paper/figures/fig_fewshot_bars.pdf`.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path("..")
FIGS = ROOT / "paper/figures"

DATA = {   # from tab:fewshot (support_sweep.json): PrecogRM vs best baseline per k
 "Tau2-Bench":  {"pn": [0.871, 0.902, 0.917, 0.927], "bl": [0.751, 0.795, 0.839, 0.875]},
 "BFCLv3":      {"pn": [0.840, 0.856, 0.876, 0.885], "bl": [0.729, 0.773, 0.811, 0.852]},
 "AgentClinic": {"pn": [0.713, 0.763, 0.784, 0.805], "bl": [0.651, 0.702, 0.728, 0.765]},
}
KS = [1, 2, 4, 8]
PN_COLOR, BL_COLOR = "#0072B2", "#b0b0b0"
YLIM = (0.55, 1.0)
FIGSIZE = (3.3125, 3.7)   # AAAI \columnwidth, include at width=\columnwidth (1:1 scale)

# ---- shared paper style: AAAI >=9pt in-figure text, Times, no Type 3 fonts ----
PAPER_RC = {
    "font.size": 9, "axes.titlesize": 10, "axes.labelsize": 10,
    "xtick.labelsize": 9, "ytick.labelsize": 9, "legend.fontsize": 9,
    "font.family": "serif",
    "font.serif": ["Times New Roman", "Times", "Nimbus Roman", "DejaVu Serif"],
    "mathtext.fontset": "stix",
    "pdf.fonttype": 42, "ps.fonttype": 42,
    "figure.dpi": 250,
}
plt.rcParams.update(PAPER_RC)
plt.rcParams.update({
    "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.alpha": 0.25, "grid.linewidth": 0.5,
})
```

```python
fig, axes = plt.subplots(3, 1, figsize=FIGSIZE, sharex=True)
x = np.arange(len(KS)); w = 0.38
for ax, (name, d) in zip(axes, DATA.items()):
    ax.bar(x - w/2, d["pn"], w, color=PN_COLOR, label="PrecogRM")
    ax.bar(x + w/2, d["bl"], w, color=BL_COLOR, label="Best per-task baseline")
    ax.axhline(d["bl"][-1], color="#444444", lw=0.8, ls="--", zorder=0)
    ax.text(0.02, 0.94, name, transform=ax.transAxes, fontsize=9,
            fontstyle="italic", va="top")
    ax.set_ylim(*YLIM)
    ax.set_yticks([0.6, 0.7, 0.8, 0.9, 1.0])
    ax.grid(axis="x", visible=False)
    ax.tick_params(axis="x", length=0)
axes[1].set_ylabel("AUROC", labelpad=5)
axes[-1].set_xticks(x); axes[-1].set_xticklabels([str(k) for k in KS])
axes[-1].set_xlabel("Support size $k$ (rollouts per class)")
axes[2].legend(frameon=False, loc="upper right",
               borderaxespad=0.1, handlelength=1.2, labelspacing=0.25,
               handletextpad=0.4)
fig.subplots_adjust(left=0.16, right=0.99, top=0.975, bottom=0.115, hspace=0.14)
FIGS.mkdir(parents=True, exist_ok=True)
fig.savefig(FIGS / "fig_fewshot_bars.pdf")
fig.savefig(FIGS / "fig_fewshot_bars.png", dpi=200)
print("saved", FIGS / "fig_fewshot_bars.{pdf,png}")
```
